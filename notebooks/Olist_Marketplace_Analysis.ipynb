Clean and Prepare the orders Dataset

```python
# Convert order date columns to datetime

order_date_cols = [
    'order_purchase_timestamp',
    'order_approved_at',
    'order_delivered_carrier_date',
    'order_delivered_customer_date',
    'order_estimated_delivery_date'
]

for col in order_date_cols:
    orders_clean[col] = pd.to_datetime(orders_clean[col])

print("Date columns converted successfully! ✅")

# Check data types
orders_clean.dtypes
```

Create Delivery & Time Features

```python
# Create time-based features

# Purchase date components
orders_clean['purchase_year'] = (
    orders_clean['order_purchase_timestamp'].dt.year
)

orders_clean['purchase_month'] = (
    orders_clean['order_purchase_timestamp'].dt.month
)

orders_clean['purchase_month_name'] = (
    orders_clean['order_purchase_timestamp'].dt.strftime('%b')
)

orders_clean['purchase_day'] = (
    orders_clean['order_purchase_timestamp'].dt.day
)

orders_clean['purchase_dayofweek'] = (
    orders_clean['order_purchase_timestamp'].dt.day_name()
)


# Delivery duration in days
orders_clean['delivery_days'] = (
    orders_clean['order_delivered_customer_date']
    - orders_clean['order_purchase_timestamp']
).dt.total_seconds() / (60 * 60 * 24)


# Delivery delay compared with estimated date
# Positive = delivered after estimated date
# Negative = delivered before estimated date

orders_clean['delivery_delay_days'] = (
    orders_clean['order_delivered_customer_date']
    - orders_clean['order_estimated_delivery_date']
).dt.total_seconds() / (60 * 60 * 24)


print("Time and delivery features created successfully! ✅")

display(
    orders_clean[
        [
            'order_purchase_timestamp',
            'order_status',
            'delivery_days',
            'delivery_delay_days'
        ]
    ].head()
)
```

Create Delivery Performance Flags

```python
# Create delivery status categories

orders_clean['delivery_performance'] = np.where(
    orders_clean['order_status'] != 'delivered',
    'Not Delivered',
    np.where(
        orders_clean['delivery_delay_days'] > 0,
        'Late',
        'On Time / Early'
    )
)

# Create a simple late-delivery flag
orders_clean['is_late'] = np.where(
    (orders_clean['order_status'] == 'delivered') &
    (orders_clean['delivery_delay_days'] > 0),
    1,
    0
)

print("Delivery performance flags created successfully! ✅")

# Check distribution
display(
    orders_clean['delivery_performance']
    .value_counts()
    .reset_index()
)
```

Validate Delivery Metrics

```python
# Validate delivery metrics

print("Negative delivery days:")
print((orders_clean['delivery_days'] < 0).sum())

print("\nDelivered orders with missing delivery_days:")
print(
    orders_clean[
        orders_clean['order_status'] == 'delivered'
    ]['delivery_days'].isna().sum()
)

print("\nDelivery days summary for delivered orders:")

display(
    orders_clean[
        orders_clean['order_status'] == 'delivered'
    ]['delivery_days'].describe()
)

print("\nLate delivery percentage:")

late_rate = (
    orders_clean[
        orders_clean['order_status'] == 'delivered'
    ]['is_late'].mean() * 100
)

print(f"{late_rate:.2f}%")
```

Validate Order Value vs Payment Value

```python
# Merge order item summary with payment summary

financial_validation = order_items_summary.merge(
    payment_summary[
        ['order_id', 'total_payment_value']
    ],
    on='order_id',
    how='inner'
)

# Calculate difference

financial_validation['payment_difference'] = (
    financial_validation['total_payment_value']
    - financial_validation['total_item_value']
)

# Round small floating-point differences
financial_validation['payment_difference'] = (
    financial_validation['payment_difference'].round(2)
)

print("Orders compared:", len(financial_validation))

print("\nOrders where payment exactly matches item total:")
print(
    (financial_validation['payment_difference'] == 0).sum()
)

print("\nOrders with differences:")
print(
    (financial_validation['payment_difference'] != 0).sum()
)

print("\nPayment difference summary:")
display(
    financial_validation['payment_difference'].describe()
)

print("\nSample orders with differences:")
display(
    financial_validation[
        financial_validation['payment_difference'] != 0
    ].head(10)
)
```

lassify Financial Differences

```python
# Classify payment differences by size

financial_validation['abs_payment_difference'] = (
    financial_validation['payment_difference'].abs()
)

# Difference categories
financial_validation['difference_category'] = pd.cut(
    financial_validation['abs_payment_difference'],
    bins=[-0.01, 0.01, 1, 10, 50, float('inf')],
    labels=[
        'Exact / Rounding (<= 0.01)',
        'Small (>0.01 to 1)',
        'Moderate (>1 to 10)',
        'Large (>10 to 50)',
        'Very Large (>50)'
    ]
)

difference_summary = (
    financial_validation['difference_category']
    .value_counts()
    .sort_index()
    .reset_index()
)

difference_summary.columns = [
    'difference_category',
    'order_count'
]

difference_summary['percentage'] = (
    difference_summary['order_count']
    / len(financial_validation) * 100
).round(3)

display(difference_summary)


# Show only material differences (> 1)

material_differences = financial_validation[
    financial_validation['abs_payment_difference'] > 1
].copy()

print("Material differences (> 1):", len(material_differences))

display(
    material_differences[
        [
            'order_id',
            'total_product_value',
            'total_freight_value',
            'total_item_value',
            'total_payment_value',
            'payment_difference',
            'abs_payment_difference'
        ]
    ].head(15)
)
```

Create Customer Satisfaction Categories

```python
# Create customer satisfaction categories

def satisfaction_category(score):

    if score <= 2:
        return 'Dissatisfied'

    elif score == 3:
        return 'Neutral'

    else:
        return 'Satisfied'


order_review_clean['satisfaction_category'] = (
    order_review_clean['review_score']
    .apply(satisfaction_category)
)

# Check distribution

satisfaction_summary = (
    order_review_clean['satisfaction_category']
    .value_counts()
    .reset_index()
)

satisfaction_summary.columns = [
    'satisfaction_category',
    'review_count'
]

satisfaction_summary['percentage'] = (
    satisfaction_summary['review_count']
    / satisfaction_summary['review_count'].sum()
    * 100
).round(2)

display(satisfaction_summary)
```

```python
# Analyze missing review comments by review score

comment_analysis = (
    order_review_clean
    .assign(
        has_title=order_review_clean['review_comment_title'].notna(),
        has_message=order_review_clean['review_comment_message'].notna()
    )
    .groupby('review_score')
    .agg(
        total_reviews=('review_score', 'count'),
        reviews_with_title=('has_title', 'sum'),
        reviews_with_message=('has_message', 'sum')
    )
    .reset_index()
)

# Calculate percentages
comment_analysis['title_percentage'] = (
    comment_analysis['reviews_with_title']
    / comment_analysis['total_reviews']
    * 100
).round(2)

comment_analysis['message_percentage'] = (
    comment_analysis['reviews_with_message']
    / comment_analysis['total_reviews']
    * 100
).round(2)

display(comment_analysis)
```

Create Review Comment Availability Flags

```python
# Create review text availability flags

order_review_clean['has_review_title'] = (
    order_review_clean['review_comment_title']
    .notna()
)

order_review_clean['has_review_message'] = (
    order_review_clean['review_comment_message']
    .notna()
)

# Create a simple review text status

def review_text_status(row):

    if row['has_review_title'] and row['has_review_message']:
        return 'Title and Message'

    elif row['has_review_message']:
        return 'Message Only'

    elif row['has_review_title']:
        return 'Title Only'

    else:
        return 'Rating Only'


order_review_clean['review_text_status'] = (
    order_review_clean.apply(review_text_status, axis=1)
)

# Check distribution

review_text_summary = (
    order_review_clean['review_text_status']
    .value_counts()
    .reset_index()
)

review_text_summary.columns = [
    'review_text_status',
    'review_count'
]

review_text_summary['percentage'] = (
    review_text_summary['review_count']
    / review_text_summary['review_count'].sum()
    * 100
).round(2)

display(review_text_summary)
```

```python
# Sort reviews by creation date

reviews_sorted = order_review_clean.sort_values(
    by=['order_id', 'review_creation_date']
)

# Get latest review for each order

latest_reviews = (
    reviews_sorted
    .groupby('order_id')
    .tail(1)[
        [
            'order_id',
            'review_score',
            'satisfaction_category',
            'review_text_status',
            'review_creation_date'
        ]
    ]
    .rename(columns={
        'review_score': 'latest_review_score',
        'satisfaction_category': 'latest_satisfaction_category',
        'review_text_status': 'latest_review_text_status',
        'review_creation_date': 'latest_review_date'
    })
)

# Create aggregated review summary

order_review_summary = (
    order_review_clean
    .groupby('order_id')
    .agg(
        average_review_score=('review_score', 'mean'),
        review_count=('review_score', 'count'),
        has_any_review_message=('has_review_message', 'max')
    )
    .reset_index()
)

# Merge latest review information

order_review_summary = order_review_summary.merge(
    latest_reviews,
    on='order_id',
    how='left'
)

print("Final review summary shape:",
      order_review_summary.shape)

print("Duplicate order IDs:",
      order_review_summary['order_id'].duplicated().sum())

display(order_review_summary.head())
```

Create Customer Value Segments

```python
# STEP 65 — Customer Value Segmentation

# Create revenue quartiles
q1 = customer_summary['total_revenue'].quantile(0.25)
q3 = customer_summary['total_revenue'].quantile(0.75)

print(f"25th Percentile Revenue: R$ {q1:.2f}")
print(f"75th Percentile Revenue: R$ {q3:.2f}")


# Function to classify customers
def customer_segment(revenue):

    if revenue <= q1:
        return 'Low Value'

    elif revenue <= q3:
        return 'Medium Value'

    else:
        return 'High Value'


customer_summary['customer_segment'] = (
    customer_summary['total_revenue']
    .apply(customer_segment)
)


# Segment summary
segment_summary = (
    customer_summary
    .groupby('customer_segment')
    .agg(
        customers=('customer_unique_id', 'count'),
        total_revenue=('total_revenue', 'sum'),
        average_customer_revenue=('total_revenue', 'mean'),
        average_orders=('total_orders', 'mean')
    )
    .reset_index()
)

# Calculate percentages
segment_summary['customer_percentage'] = (
    segment_summary['customers']
    / segment_summary['customers'].sum()
    * 100
)

segment_summary['revenue_percentage'] = (
    segment_summary['total_revenue']
    / segment_summary['total_revenue'].sum()
    * 100
)

print("\nCUSTOMER SEGMENT SUMMARY")
print("=" * 60)

display(segment_summary.round(2))
```

🎯 BUSINESS FINDINGS & ACTIONABLE RECOMMENDATIONS
======================================================================

1️⃣ CUSTOMER RETENTION PROBLEM

Finding:
97% of customers are one-time buyers, while repeat customers generate
significantly higher revenue per customer.

Business Impact:
The company has a major opportunity to increase Customer Lifetime Value (CLV)
by converting one-time buyers into repeat customers.

Recommendation:
• Launch second-purchase campaigns
• Introduce loyalty programs
• Send personalized product recommendations
• Create customer reactivation campaigns


2️⃣ DELIVERY DELAYS REDUCE CUSTOMER SATISFACTION

Finding:
Late deliveries have an average review score of 2.55 compared with 4.28
for on-time or early deliveries.

Business Impact:
Poor delivery performance directly affects customer satisfaction and may
negatively impact customer retention.

Recommendation:
• Implement delivery delay monitoring
• Create early-warning systems for delayed orders
• Improve logistics in high-delay states
• Monitor seller dispatch performance


3️⃣ HIGH-VALUE CUSTOMERS ARE CRITICAL

Finding:
25% of High-Value customers generate approximately 59.43% of total revenue.

Business Impact:
Losing High-Value customers could significantly affect overall business revenue.

Recommendation:
• Create a VIP customer program
• Provide personalized offers
• Offer exclusive benefits
• Prioritize High-Value customer retention


4️⃣ GEOGRAPHIC REVENUE CONCENTRATION

Finding:
São Paulo generates approximately 37.41% of customer revenue, making it
the company's largest market.

Business Impact:
Heavy dependency on one geographic market creates concentration risk.

Recommendation:
• Expand strategically into high-potential states
• Improve logistics infrastructure outside São Paulo
• Increase seller availability in growth markets


5️⃣ SELLER REVENUE CONCENTRATION

Finding:
Approximately 25% of High-Value sellers generate 86.58% of total seller revenue.

Business Impact:
The platform is highly dependent on a relatively small group of sellers.

Recommendation:
• Retain and support High-Value sellers
• Develop Medium-Value sellers
• Improve seller performance programs
• Reduce dependency on a small number of top sellers

The biggest opportunity for sustainable business growth is not only acquiring
new customers, but improving customer retention.

The company should focus on:

1. Increasing repeat purchases
2. Reducing delivery delays
3. Protecting High-Value customers
4. Supporting High-Value and Medium-Value sellers
5. Expanding strategically beyond São Paulo
